## 🎯 Project Objective

The GenAI Student & Career Assistant is an AI-powered
Generative AI application that combines:

tools, and generate useful responses.

---

## 🔄 Overall Workflow

The GenAI Student & Career Assistant follows this workflow:

### Step 1 — 👤 User Request
The user enters a question, task, or request.

⬇️

### Step 2 — 🧠 Understand Request
The system analyzes the user's intent and determines what type of response is required.

⬇️

### Step 3 — ✨ Prompt Processing
The request is processed using structured prompts and appropriate instructions.

⬇️

### Step 4 — 🤖 AI Agent Decision
The AI agent decides whether to:

- 💬 Generate a direct LLM response
- 📚 Retrieve information from uploaded documents using RAG
- 🛠️ Execute an available tool

⬇️

### Step 5 — ⚙️ Action

| Option | Action |
|---|---|
| 💬 LLM | Generate an AI response |
| 📚 RAG | Retrieve relevant document context |
| 🛠️ Tool | Execute the required function |

⬇️

### Step 6 — 📊 Process Results
The retrieved information or tool results are processed by the AI system.

⬇️

### Step 7 — 🤖 Generate Final Response
The LLM combines the request, retrieved context, and tool results when required.

⬇️

### Step 8 — 💬 User Response
The final answer is presented to the user through the application.

---

### 🔁 Complete Pipeline

**User Request → Understand → Prompt Processing → Agent Decision → LLM / RAG / Tools → Process Results → Final Response**

```python
# ============================================
# STEP 2: LLM API CONNECTION
# ============================================

!pip -q install huggingface_hub

from google.colab import userdata
from huggingface_hub import InferenceClient

# Get Hugging Face token securely from Colab Secrets
HF_TOKEN = userdata.get("HF_TOKEN")

# Create Hugging Face client
client = InferenceClient(api_key=HF_TOKEN)

# LLM model
MODEL = "openai/gpt-oss-120b:fastest"

print("✅ Hugging Face LLM client connected successfully!")
print("🤖 Model:", MODEL)
```

```python
# ============================================
# STEP 3: BASIC LLM RESPONSE
# ============================================

response = client.chat.completions.create(
    model=MODEL,
    messages=[
        {
            "role": "system",
            "content": """You are a helpful Student and Career Assistant.
Give clear, practical, and beginner-friendly answers.
"""
        },
        {
            "role": "user",
            "content": "Give me 3 practical tips for preparing for an AI/ML career."
        }
    ],
    max_tokens=250
)

print("🤖 AI Assistant Response:\n")
print(response.choices[0].message.content)
```

```python
# ============================================
# STEP 4: PROMPT ENGINEERING
# ============================================

SYSTEM_PROMPT = """
You are a professional GenAI Student & Career Assistant.

Your job is to help users with:
1. 📚 Study and learning
2. 💼 Career guidance
3. 💻 Programming and coding
4. 📄 Document-related questions
5. 🤖 General AI and technology questions

Response guidelines:

- Understand the user's intent before answering.
- Give clear, practical, and structured responses.
- Use simple language when explaining difficult concepts.
- For study questions, provide step-by-step explanations.
- For career questions, provide actionable recommendations.
- For coding questions, explain the logic before giving code.
- For document questions, rely on retrieved document context when available.
- Never invent information that is not available in the provided context.
- If the required information is unavailable, clearly say so.
- Use headings, bullet points, and numbered steps when useful.
"""

response = client.chat.completions.create(
    model=MODEL,
    messages=[
        {
            "role": "system",
            "content": SYSTEM_PROMPT
        },
        {
            "role": "user",
            "content": "Explain what Retrieval-Augmented Generation (RAG) is to a beginner."
        }
    ],
    max_tokens=300
)

print("🧠 Prompt-Engineered AI Response:\n")
print(response.choices[0].message.content)
```

```python
# ============================================
# STEP 5A: RAG LIBRARIES
# ============================================

!pip -q install sentence-transformers chromadb

from sentence_transformers import SentenceTransformer
import chromadb

# Embedding model
embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

# Vector database
chroma_client = chromadb.Client()

# Create collection
collection = chroma_client.get_or_create_collection(
    name="student_career_knowledge"
)

print("✅ RAG libraries loaded successfully!")
print("✅ Embedding model loaded!")
print("✅ ChromaDB collection created!")
```

```python
# ============================================
# STEP 5B: CREATE KNOWLEDGE BASE
# ============================================

documents = [
    {
        "id": "study_1",
        "text": """
        Effective study planning starts by identifying learning goals,
        breaking large subjects into smaller topics, and creating a
        realistic study schedule. Active recall, spaced repetition,
        practice questions, and regular revision can improve learning.
        """
    },
    {
        "id": "career_1",
        "text": """
        Building an AI and machine learning career requires strong
        programming fundamentals, knowledge of mathematics and machine
        learning concepts, practical project experience, and a portfolio.
        GitHub projects, internships, technical practice, and continuous
        learning can help demonstrate practical skills.
        """
    },
    {
        "id": "coding_1",
        "text": """
        Good programming practice involves understanding the problem,
        designing an algorithm, writing readable code, testing the solution,
        handling errors, and improving the solution when necessary.
        Python is widely used in artificial intelligence and machine learning.
        """
    },
    {
        "id": "rag_1",
        "text": """
        Retrieval-Augmented Generation, or RAG, combines information
        retrieval with a large language model. Relevant information is
        retrieved from a knowledge source and provided as context to the
        language model before generating an answer. RAG can help ground
        responses in specific information.
        """
    },
    {
        "id": "genai_1",
        "text": """
        Generative AI systems use machine learning models to generate
        content such as text, code, images, or other outputs. Large
        language models can understand and generate natural language
        based on patterns learned during training.
        """
    }
]

# Extract text and IDs
texts = [doc["text"] for doc in documents]
ids = [doc["id"] for doc in documents]

# Create embeddings
embeddings = embedding_model.encode(texts).tolist()

# Store documents and embeddings in ChromaDB
collection.add(
    ids=ids,
    documents=texts,
    embeddings=embeddings
)

print("✅ Knowledge base created!")
print(f"📚 Documents added: {len(documents)}")
print(f"🔢 Embedding dimensions: {len(embeddings[0])}")
print("🗄️ Documents stored in ChromaDB!")
```

```python
# ============================================
# STEP 5C: SEMANTIC RETRIEVAL
# ============================================

def retrieve_context(query, top_k=2):
    # Convert the user query into an embedding
    query_embedding = embedding_model.encode([query]).tolist()[0]

    # Search ChromaDB for the most relevant documents
    results = collection.query(
        query_embeddings=[query_embedding],
        n_results=top_k
    )

    retrieved_documents = results["documents"][0]

    return retrieved_documents


# Test the retrieval system
query = "How can I prepare for an AI and machine learning career?"

retrieved_context = retrieve_context(query)

print("🔍 User Query:")
print(query)

print("\n📚 Retrieved Context:\n")

for i, context in enumerate(retrieved_context, start=1):
    print(f"--- Relevant Document {i} ---")
    print(context.strip())
    print()
```

```python
# ============================================
# STEP 5D: RAG + LLM GENERATION
# ============================================

def rag_answer(user_query):
    # Retrieve relevant information
    retrieved_context = retrieve_context(user_query, top_k=2)

    # Combine retrieved documents
    context = "\n\n".join(retrieved_context)

    # Create RAG prompt
    rag_prompt = f"""
You are a helpful Student & Career Assistant.

Answer the user's question using the provided context.

IMPORTANT:
- Use the context as the primary source of information.
- Do not invent facts that are not supported by the context.
- If the context does not contain enough information, clearly say so.
- Give a clear and practical answer.

CONTEXT:
{context}

USER QUESTION:
{user_query}
"""

    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {
                "role": "system",
                "content": SYSTEM_PROMPT
            },
            {
                "role": "user",
                "content": rag_prompt
            }
        ],
        max_tokens=350
    )

    return response.choices[0].message.content


# Test the complete RAG pipeline
question = "What should I do to build a career in AI and machine learning?"

answer = rag_answer(question)

print("👤 User Question:")
print(question)

print("\n🤖 RAG AI Assistant Answer:\n")
print(answer)
```

```python
# ============================================
# STEP 6: AI AGENT TOOLS
# ============================================

import json

# --------------------------------------------
# 🧮 Calculator Tool
# --------------------------------------------

def calculator(expression):
    try:
        result = eval(expression, {"__builtins__": {}}, {})
        return f"Calculation result: {result}"
    except Exception:
        return "Unable to calculate the expression."


# --------------------------------------------
# 📝 Task Manager
# --------------------------------------------

tasks = []


def add_task(task):
    tasks.append(task)
    return f"Task added successfully: {task}"


def list_tasks():
    if not tasks:
        return "No tasks found."

    result = "Current tasks:\n"

    for i, task in enumerate(tasks, start=1):
        result += f"{i}. {task}\n"

    return result


def remove_task(task_number):
    try:
        task_number = int(task_number)

        if 1 <= task_number <= len(tasks):
            removed = tasks.pop(task_number - 1)
            return f"Task removed: {removed}"

        return "Invalid task number."

    except Exception:
        return "Invalid task number."


print("🛠️ AI tools initialized successfully!")
print("🧮 Calculator tool ready")
print("📝 Task manager ready")
```

```python
# ============================================
# STEP 7: AI AGENT TOOL DEFINITIONS
# ============================================

tools = [
    {
        "type": "function",
        "function": {
            "name": "calculator",
            "description": "Calculate a mathematical expression.",
            "parameters": {
                "type": "object",
                "properties": {
                    "expression": {
                        "type": "string",
                        "description": "Mathematical expression such as 125 * 32"
                    }
                },
                "required": ["expression"]
            }
        }
    },
    {
        "type": "function",
        "function": {
            "name": "add_task",
            "description": "Add a new task to the user's task list.",
            "parameters": {
                "type": "object",
                "properties": {
                    "task": {
                        "type": "string",
                        "description": "The task to add"
                    }
                },
                "required": ["task"]
            }
        }
    },
    {
        "type": "function",
        "function": {
            "name": "list_tasks",
            "description": "Show all tasks currently stored.",
            "parameters": {
                "type": "object",
                "properties": {}
            }
        }
    },
    {
        "type": "function",
        "function": {
            "name": "remove_task",
            "description": "Remove a task using its task number.",
            "parameters": {
                "type": "object",
                "properties": {
                    "task_number": {
                        "type": "integer",
                        "description": "The number of the task to remove"
                    }
                },
                "required": ["task_number"]
            }
        }
    },
    {
        "type": "function",
        "function": {
            "name": "search_knowledge",
            "description": "Search the internal knowledge base for study, career, coding, RAG, and GenAI information.",
            "parameters": {
                "type": "object",
                "properties": {
                    "query": {
                        "type": "string",
                        "description": "Question or topic to search for"
                    }
                },
                "required": ["query"]
            }
        }
    }
]

print("✅ AI agent tool definitions created!")
print(f"🛠️ Available tools: {len(tools)}")
```

```python
# ============================================
# STEP 8: TOOL EXECUTION + AI AGENT
# ============================================

def search_knowledge(query):
    """Search the ChromaDB knowledge base."""
    retrieved = retrieve_context(query, top_k=2)

    if not retrieved:
        return "No relevant information found."

    return "\n\n".join(retrieved)


def execute_tool(tool_name, arguments):
    """Execute the tool selected by the AI agent."""

    if tool_name == "calculator":
        return calculator(arguments.get("expression", ""))

    elif tool_name == "add_task":
        return add_task(arguments.get("task", ""))

    elif tool_name == "list_tasks":
        return list_tasks()

    elif tool_name == "remove_task":
        return remove_task(arguments.get("task_number"))

    elif tool_name == "search_knowledge":
        return search_knowledge(arguments.get("query", ""))

    return "Unknown tool."


def run_agent(user_request, max_steps=5):

    messages = [
        {
            "role": "system",
            "content": SYSTEM_PROMPT + """

You are also an AI agent.

You have access to:
- Calculator
- Task Manager
- Knowledge Base Search

IMPORTANT RULES:
1. Use a tool whenever the user's request requires an action.
2. Use the knowledge search tool for questions related to
   study, career, coding, RAG, or GenAI when relevant.
3. Never claim that a tool action was completed unless the tool
   actually executed successfully.
4. If multiple actions are required, continue until all required
   actions are completed.
5. After receiving a tool result, decide whether another tool
   is required.
6. Only provide the final answer after completing the required actions.
"""
        },
        {
            "role": "user",
            "content": user_request
        }
    ]

    activity = []

    for step in range(max_steps):

        response = client.chat.completions.create(
            model=MODEL,
            messages=messages,
            tools=tools,
            tool_choice="auto",
            max_tokens=400
        )

        assistant_message = response.choices[0].message

        # No more tools required
        if not getattr(assistant_message, "tool_calls", None):
            return {
                "answer": assistant_message.content,
                "activity": activity
            }

        # Add assistant tool-call message
        messages.append(assistant_message)

        # Execute every requested tool
        for tool_call in assistant_message.tool_calls:

            tool_name = tool_call.function.name

            arguments = json.loads(
                tool_call.function.arguments
            )

            activity.append(
                f"🛠️ Executing {tool_name}: {arguments}"
            )

            tool_result = execute_tool(
                tool_name,
                arguments
            )

            activity.append(
                f"📤 Result: {tool_result}"
            )

            messages.append(
                {
                    "role": "tool",
                    "tool_call_id": tool_call.id,
                    "content": str(tool_result)
                }
            )

    return {
        "answer": "The agent reached the maximum number of steps.",
        "activity": activity
    }


print("🤖 AI Agent initialized successfully!")
print("🛠️ Tool execution system ready!")
print("📚 RAG knowledge search connected!")
print("🔄 Multi-step agent workflow ready!")
```

```python
# ============================================
# STEP 9: TEST THE COMPLETE AI AGENT
# ============================================

test_requests = [
    "What skills are important for building a career in AI and machine learning?",
    "Calculate 125 * 32",
    "Add a task called Complete my GenAI capstone project"
]

for i, request in enumerate(test_requests, start=1):

    print("=" * 60)
    print(f"🧪 TEST {i}")
    print("=" * 60)

    print(f"\n👤 User Request:")
    print(request)

    result = run_agent(request)

    print("\n🤖 Agent Response:")
    print(result["answer"])

    print("\n🛠️ Agent Activity:")
    for activity in result["activity"]:
        print(activity)

    print()
```
